## 1. Variablen und Datentypen

Eine **Variable** speichert einen Wert unter einem Namen. Das Gleichheitszeichen ist eine **Zuweisung**: rechts wird berechnet, links unter einem Namen gespeichert.

````python
preis = 20
menge = 3
umsatz = preis * menge
print(umsatz)
````

**Namen** bestehen aus Kleinbuchstaben, Ziffern und `_` und beginnen nicht mit einer Ziffer, z. B. `total_price`. Groß-/Kleinschreibung zählt: `umsatz` und `Umsatz` sind zwei verschiedene Variablen.

### Reihenfolge der Anweisungen

Python führt Anweisungen **der Reihe nach** aus und merkt sich den aktuellen Wert jeder Variable (den **Zustand**). Eine Variable behält ihren Wert, bis ihr ein neuer zugewiesen wird. Sie rechnet sich **nicht** von selbst neu, wenn sich eine andere Variable ändert:

````python
preis = 20
menge = 3
umsatz = preis * menge
menge = 5
print(umsatz)    # immer noch 60: berechnet, als menge noch 3 war

umsatz = preis * menge
print(umsatz)    # jetzt 100
````

Der Zustand nach jedem Schritt (**fett**: neuer Wert):

| Schritt | Anweisung | `preis` | `menge` | `umsatz` |
|---|---|---|---|---|
| 1 | `preis = 20` | **20** | | |
| 2 | `menge = 3` | 20 | **3** | |
| 3 | `umsatz = preis * menge` | 20 | 3 | **60** |
| 4 | `menge = 5` | 20 | **5** | 60 |
| 5 | `umsatz = preis * menge` | 20 | 5 | **100** |

In Colab gilt dasselbe für Zellen: Maßgeblich ist die Reihenfolge, in der die Zellen **ausgeführt** werden, nicht die Reihenfolge im Notebook.

### Vor den Aufgaben: Tests vorbereiten

Jede Aufgabe hat eine **Testzelle**, die das Ergebnis prüft. Dafür einmal die folgende Zelle **Tests vorbereiten** ausführen. Ihr Inhalt ist für die Aufgaben nicht wichtig. Nach einem Neustart der Sitzung muss sie erneut ausgeführt werden.

````python
#@title ▶ Tests vorbereiten (vor den Aufgaben einmal ausführen)
def _gleich(a, b):
    if isinstance(a, bool) or isinstance(b, bool):
        return a is b
    if isinstance(a, float) or isinstance(b, float):
        try:
            return abs(a - b) < 1e-6
        except TypeError:
            return False
    if isinstance(a, (list, tuple)) and isinstance(b, (list, tuple)):
        return len(a) == len(b) and all(_gleich(x, y) for x, y in zip(a, b))
    return a == b

def pruefe(tatsaechlich, erwartet):
    """Vergleicht ein Ergebnis mit dem erwarteten Wert."""
    if _gleich(tatsaechlich, erwartet):
        print("✅ Richtig!")
    else:
        print("❌ Noch nicht richtig.")
        print("   erwartet:", repr(erwartet))
        print("   erhalten:", repr(tatsaechlich))

print("Tests bereit.")
````

## 2. Vergleiche und Logik

Vergleiche liefern `True` oder `False`. Vieles ist aus SQL bekannt, einiges wird anders geschrieben:

| SQL | Python | Bedeutung |
|---|---|---|
| `=` | `==` | gleich |
| `<>` | `!=` | ungleich |
| `<`, `>`, `<=`, `>=` | gleich wie in SQL | kleiner, größer … |
| `AND`, `OR`, `NOT` | `and`, `or`, `not` | verknüpfen |
| `IN ('A', 'B')` | `in ["A", "B"]` | ist in der Liste |

Achtung: Ein einfaches `=` ist in Python eine Zuweisung, kein Vergleich.

````python
preis = 53
land = "Austria"

print(preis > 50)
print(land == "Germany")
print(land != "Germany")
print(preis > 50 and land == "Austria")
print(land in ["Austria", "Germany", "Switzerland"])
````

Anders als im SQL-Editor zählt in Python die **Groß-/Kleinschreibung**:

````python
print("Austria" == "austria")
````

## 3. Listen und Dictionaries

Eine **Liste** speichert mehrere Werte in fester Reihenfolge.

````python
bestellwerte = [120.5, 89.9, 230.0, 45.0]

print(bestellwerte[0])     # 120.5 (Index ab 0)
print(bestellwerte[-1])    # 45.0 (vom Ende)
print(len(bestellwerte))   # 4
````

### Zugriff über den Index

Jeder Wert hat zwei Indizes: **von vorne** ab `0` und **von hinten** ab `-1`.

| Index von vorne | `0` | `1` | `2` | `3` |
|---|---|---|---|---|
| **Wert** | 120.5 | 89.9 | 230.0 | 45.0 |
| **Index von hinten** | `-4` | `-3` | `-2` | `-1` |

`append()` hängt einen Wert hinten an. Danach verschieben sich die Indizes von hinten: `99.0` ist jetzt `[-1]`, `45.0` ist `[-2]`.

````python
bestellwerte = [120.5, 89.9, 230.0, 45.0]
bestellwerte.append(99.0)   # hinten anhängen
print(bestellwerte)
print(bestellwerte[-1], bestellwerte[-2])
print(max(bestellwerte))
````

Weitere Funktionen für Listen: `sum`, `min`, `max` und `sorted`.

````python
bestellwerte = [120.5, 89.9, 230.0, 45.0]
print(sum(bestellwerte), min(bestellwerte), max(bestellwerte))
print(sorted(bestellwerte))                 # aufsteigend
print(sorted(bestellwerte, reverse=True))   # absteigend
````

## 4. Verzweigungen: if, elif, else

Mit `if` wird Code nur ausgeführt, wenn eine Bedingung erfüllt ist. `elif` prüft weitere Bedingungen, `else` greift, wenn keine zutrifft. Geprüft wird von oben nach unten, die erste zutreffende Bedingung gewinnt.

**Wichtig:** Nach der Bedingung steht ein **Doppelpunkt**, der abhängige Code ist **eingerückt** (4 Leerzeichen).

````python
bestellwert = 180

if bestellwert >= 200:
    rabatt = 0.10
elif bestellwert >= 100:
    rabatt = 0.05
else:
    rabatt = 0

print("Rabatt:", rabatt * 100, "%")
````

Aufbau: `def`, Name, Parameter in Klammern, Doppelpunkt, eingerückter Block.

### Funktionen testen

Funktionen werden mit **Testfällen** geprüft: eine Eingabe, das erwartete Ergebnis und das tatsächliche Ergebnis. Sinnvoll sind mehrere Testfälle, auch **Grenzfälle** wie genau 50 oder eine leere Liste. Die folgende Funktion `check_result` ist dieselbe wie in der Gruppenübung:

````python
def check_result(i, expected_result, actual_result):
  print("Input: " + str(i))
  print("Expected result: " + str(expected_result))
  print("Actual result: " + str(actual_result))
  if expected_result == actual_result:
    print("✅ OK! (actual result == expected result)")
  else:
    print("❌ ERROR! (actual result != expected result)")
  print()

check_result(100, 120.0, brutto(100))
````

## 8. SQL aus Python

In Colab lässt sich dieselbe Beispieldatenbank wie im SQL-Editor laden. Die folgende Zelle ist dieselbe wie in der Gruppenübung: Sie lädt die Daten und stellt die Datenbank in der Variable `db` bereit.

````python
import os
from urllib.request import urlopen

GITHUB_URL = "https://raw.githubusercontent.com/markusschanta/w3schools-database/refs/heads/master/w3schools.sqllite.sql"
create_sql = urlopen(GITHUB_URL).read().decode()

import sqlite3

try:
  db.close()
except:
  pass

db = sqlite3.connect("file::memory:?cache=shared")

db.executescript(create_sql)
db.commit()
````

`pd.read_sql_query` führt eine Abfrage aus und liefert das Ergebnis als **Tabelle** (pandas DataFrame, mehr dazu in Teil 4):

````python
import pandas as pd

SQL = """
SELECT Country, COUNT(*) AS Kunden
FROM customers
GROUP BY Country
"""
pd.read_sql_query(SQL, db)
````

**Unterschiede zum SQL-Editor des Kurses** (hier läuft SQLite ohne dessen Anpassungen):

* Tabellennamen klein, die Bestellpositionen heißen `order_details` (statt `OrderDetails`).
* Bei Text zählt die Groß-/Kleinschreibung: `'Austria'` findet Kunden, `'austria'` nicht.
* `TOP` gibt es nicht, nur `LIMIT` am Ende.
* Eine Spalte, die weder im `GROUP BY` noch in einer Aggregatfunktion steht, führt zu **keinem Fehler**, sondern zu einem beliebigen Wert. Die Regel aus Teil 2 also selbst beachten.

### ✏️ Aufgabe 8.2

Alle Kunden **ohne Bestellung** abfragen (Name und Land) und in `ohne_bestellung` speichern. (Tipp: `LEFT JOIN` und `IS NULL`, siehe Teil 2.)

````python
SQL = """
SELECT *
FROM customers
"""  # Abfrage hier anpassen
ohne_bestellung = pd.read_sql_query(SQL, db)
ohne_bestellung
````

````python
#@title 💡 Lösung (Doppelklick zeigt den Code)
SQL = """
SELECT c.CustomerName, c.Country
FROM customers c
LEFT JOIN orders o ON c.CustomerID = o.CustomerID
WHERE o.OrderID IS NULL
"""
ohne_bestellung = pd.read_sql_query(SQL, db)
ohne_bestellung
````
