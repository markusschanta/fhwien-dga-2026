* ```input```: Einen Eingabewert
* ```expected_result```: Den erwarteten Rückgabewert der Funktion
* ```actual_result```: Den tatsächlichen Rückgabewert der Funktion

Je nachdem, ob der tatsächliche Rückgabewert mit dem erwarteten übereinstimmt, gibt die Funktion  ```check_result``` einen anderen Wert über die Ausgabe aus.

````python
def check_result(i, expected_result, actual_result):
  print("Input: " + str(i))
  print("Expected result: " + str(expected_result))
  print("Actual result: " + str(actual_result))
  if expected_result == actual_result:
    print("✅ OK! (actual result == expected result)")
  else:
    print("❌ ERROR! (actual result != expected result)")
````
